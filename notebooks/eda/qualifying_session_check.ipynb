```python
import fastf1
```

```python
sessions = fastf1.get_session(2026, 3, 4)
sessions.load()
```

```python
import pandas as pd
all_telemetry = []
driver_laps = sessions.laps.pick_driver('LEC')
for _, lap in driver_laps.iterlaps():
    telemetry = pd.DataFrame(lap.get_telemetry())
    all_telemetry.append(telemetry)
df = pd.concat(all_telemetry, ignore_index=True)
df
```
