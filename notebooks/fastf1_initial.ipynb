```python
import os
import fastf1
from pathlib import Path
from dotenv import load_dotenv
```

```python
notebook_dir = Path.cwd()
repo_root = notebook_dir.parent
load_dotenv(repo_root / ".env")
cache_path = repo_root / os.getenv("FASTF1_CACHE_PATH", "data/fastf1_cache")
cache_path.mkdir(parents=True, exist_ok=True)

fastf1.Cache.enable_cache(str(cache_path))

session = fastf1.get_session(2026, "Australia", "R")
session.load()
print(session.event['EventName'])
print(f"{len(session.results)} drivers loaded")
print(f"fastest lap: {session.laps.pick_fastest()['Driver']}")
```
